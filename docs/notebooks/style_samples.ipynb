## 1.3  Seven ways to choose an area

*In this section: get an area from a name, an OSM id, a box, a point, a polygon or a file.*

`duckosm.get` gives an `Area`: the cut OSM file and its boundary. It does not build.

| Argument | The area | Example |
|---|---|---|
| `place` | the first answer of Nominatim for a name | `duckosm.get("Monaco")` |
| `osm_id` | an OSM relation (a city, a district) | `duckosm.get(osm_id=2220322)` |
| `bbox` | a box: west, south, east, north | `duckosm.get(bbox=(7.41, 43.72, 7.44, 43.75))` |
| `point`, `dist_m` | the box around a point | `duckosm.get(point=(7.4246, 43.7384), dist_m=500)` |

```python
casino = duckosm.get(point=(7.4280, 43.7393), dist_m=500, name="casino")   # an Area: the file and the boundary
casino
```

>
